```python
# ---------------------------------------------------------------
# Fashion-MNIST Classification using ANN and CNN
# Author: Satya Abhishek Kasukurti (P2895904)
# MSc Artificial Intelligence | Module: Neural Networks
# ---------------------------------------------------------------

import tensorflow as tf
from tensorflow.keras import datasets, models, layers
import matplotlib.pyplot as plt
import numpy as np
from sklearn.metrics import classification_report, confusion_matrix
import seaborn as sns

# ---------------------------------------------------------------
# 1. Load and Preprocess the Fashion-MNIST Dataset
# ---------------------------------------------------------------
print(f"TensorFlow version: {tf.__version__}")
print("\n--- 1. Loading and Preprocessing Data ---")

# Load dataset
(train_images, train_labels), (test_images, test_labels) = datasets.fashion_mnist.load_data()

# Class names
class_names = ['T-shirt/top', 'Trouser', 'Pullover', 'Dress', 'Coat',
               'Sandal', 'Shirt', 'Sneaker', 'Bag', 'Ankle boot']

# Normalize pixel values
train_images = train_images.astype('float32') / 255.0
test_images = test_images.astype('float32') / 255.0

# Prepare CNN data (add channel dimension)
train_images_cnn = np.expand_dims(train_images, -1)
test_images_cnn = np.expand_dims(test_images, -1)
INPUT_SHAPE_CNN = (28, 28, 1)

print(f"Train images shape (for ANN): {train_images.shape}")
print(f"Train images shape (for CNN): {train_images_cnn.shape}")

# ---------------------------------------------------------------
# 2. Artificial Neural Network (ANN/MLP) Model
# ---------------------------------------------------------------
print("\n--- 2. Training ANN (MLP) Model ---")

ann_model = models.Sequential([
    layers.Flatten(input_shape=(28, 28)),
    layers.Dense(128, activation='relu'),
    layers.Dense(64, activation='relu'),
    layers.Dense(10, activation='softmax')
])

ann_model.compile(optimizer='adam',
                  loss='sparse_categorical_crossentropy',
                  metrics=['accuracy'])

print("ANN Model Summary:")
ann_model.summary()

ann_history = ann_model.fit(
    train_images, train_labels,
    epochs=10,
    validation_split=0.1,
    verbose=1
)

# ---------------------------------------------------------------
# 3. Convolutional Neural Network (CNN) Model
# ---------------------------------------------------------------
print("\n--- 3. Training CNN Model ---")

cnn_model = models.Sequential([
    layers.Conv2D(32, (3, 3), activation='relu', input_shape=INPUT_SHAPE_CNN),
    layers.MaxPooling2D((2, 2)),
    layers.Conv2D(64, (3, 3), activation='relu'),
    layers.MaxPooling2D((2, 2)),
    layers.Flatten(),
    layers.Dense(100, activation='relu'),
    layers.Dense(10, activation='softmax')
])

cnn_model.compile(optimizer='adam',
                  loss='sparse_categorical_crossentropy',
                  metrics=['accuracy'])

print("CNN Model Summary:")
cnn_model.summary()

cnn_history = cnn_model.fit(
    train_images_cnn, train_labels,
    epochs=10,
    validation_split=0.1,
    verbose=1
)

# ---------------------------------------------------------------
# 4. Evaluation and Comparison
# ---------------------------------------------------------------
print("\n--- 4. Model Evaluation and Comparison ---")

# Evaluate ANN
ann_loss, ann_acc = ann_model.evaluate(test_images, test_labels, verbose=0)
print(f"\nANN (MLP) Final Test Accuracy: {ann_acc*100:.2f}%")
print(f"ANN (MLP) Final Test Loss: {ann_loss:.4f}")

# Evaluate CNN
cnn_loss, cnn_acc = cnn_model.evaluate(test_images_cnn, test_labels, verbose=0)
print(f"\nCNN Final Test Accuracy: {cnn_acc*100:.2f}%")
print(f"CNN Final Test Loss: {cnn_loss:.4f}")

# Predictions
ann_predictions = ann_model.predict(test_images)
cnn_predictions = cnn_model.predict(test_images_cnn)

ann_predicted_labels = np.argmax(ann_predictions, axis=1)
cnn_predicted_labels = np.argmax(cnn_predictions, axis=1)

# ---------------------------------------------------------------
# 5. Classification Reports
# ---------------------------------------------------------------
print("\n--- ANN Classification Report ---")
print(classification_report(test_labels, ann_predicted_labels, target_names=class_names))

print("\n--- CNN Classification Report ---")
print(classification_report(test_labels, cnn_predicted_labels, target_names=class_names))

# ---------------------------------------------------------------
# 6. Visualization: Separate Accuracy & Loss Curves
# ---------------------------------------------------------------

# --- ANN Accuracy ---
plt.figure(figsize=(7, 5))
plt.plot(ann_history.history['accuracy'], label='Train Accuracy', color='blue')
plt.plot(ann_history.history['val_accuracy'], label='Validation Accuracy', color='cyan')
plt.title('ANN Training vs Validation Accuracy')
plt.xlabel('Epochs')
plt.ylabel('Accuracy')
plt.legend()
plt.show()

# --- ANN Loss ---
plt.figure(figsize=(7, 5))
plt.plot(ann_history.history['loss'], label='Train Loss', color='blue')
plt.plot(ann_history.history['val_loss'], label='Validation Loss', color='cyan')
plt.title('ANN Training vs Validation Loss')
plt.xlabel('Epochs')
plt.ylabel('Loss')
plt.legend()
plt.show()

# --- CNN Accuracy ---
plt.figure(figsize=(7, 5))
plt.plot(cnn_history.history['accuracy'], label='Train Accuracy', color='red')
plt.plot(cnn_history.history['val_accuracy'], label='Validation Accuracy', color='orange')
plt.title('CNN Training vs Validation Accuracy')
plt.xlabel('Epochs')
plt.ylabel('Accuracy')
plt.legend()
plt.show()

# --- CNN Loss ---
plt.figure(figsize=(7, 5))
plt.plot(cnn_history.history['loss'], label='Train Loss', color='red')
plt.plot(cnn_history.history['val_loss'], label='Validation Loss', color='orange')
plt.title('CNN Training vs Validation Loss')
plt.xlabel('Epochs')
plt.ylabel('Loss')
plt.legend()
plt.show()

# ---------------------------------------------------------------
# 7. Confusion Matrices for ANN and CNN
# ---------------------------------------------------------------

# --- ANN Confusion Matrix ---
plt.figure(figsize=(10, 8))
cm_ann = confusion_matrix(test_labels, ann_predicted_labels)
sns.heatmap(cm_ann, annot=True, fmt='d', cmap='Blues',
            xticklabels=class_names, yticklabels=class_names)
plt.title('ANN Confusion Matrix - Fashion-MNIST')
plt.ylabel('True Label')
plt.xlabel('Predicted Label')
plt.show()

# --- CNN Confusion Matrix ---
plt.figure(figsize=(10, 8))
cm_cnn = confusion_matrix(test_labels, cnn_predicted_labels)
sns.heatmap(cm_cnn, annot=True, fmt='d', cmap='Greens',
            xticklabels=class_names, yticklabels=class_names)
plt.title('CNN Confusion Matrix - Fashion-MNIST')
plt.ylabel('True Label')
plt.xlabel('Predicted Label')
plt.show()

print("\n✅ Code Execution Completed Successfully!")
```
